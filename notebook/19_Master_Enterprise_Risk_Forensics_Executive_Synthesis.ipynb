## 3. Phase 3 & 4 Synthesis: Risk Scoring, Asymmetric Cost Optimization, & Feature Store (Notebooks 07–10)

### Mathematical Formulations:
1. **Weight of Evidence (WoE) & Information Value (IV)**:
   $$\text{WoE}_i = \ln\left( \frac{\text{Good}_i / \text{Good}_{\text{total}}}{\text{Bad}_i / \text{Bad}_{\text{total}}} \right), \quad \text{IV} = \sum_{i=1}^k \left( \frac{\text{Good}_i}{\text{Good}_{\text{total}}} - \frac{\text{Bad}_i}{\text{Bad}_{\text{total}}} \right) \cdot \text{WoE}_i$$
2. **Asymmetric Financial Risk Matrix & Optimal Policy Search**:
   $$\mathbb{E}[\text{Cost}(\theta)] = \sum_{i=1}^N \Big[ y_i (1 - \hat{y}_i(\theta)) (\text{Amount}_i + C_{\text{chargeback}}) + (1 - y_i) \hat{y}_i(\theta) (C_{\text{friction}} + C_{\text{review}}) \Big]$$
   $$\theta^* = \arg\min_{\theta \in (0, 1)} \mathbb{E}[\text{Cost}(\theta)]$$
3. **Out-of-Time (OOT) Leak-Free Feature Store Splitting**:
   $$\mathcal{D}_{\text{train}} = \{ (x_t, y_t) \mid t \le T_1 \}, \quad \mathcal{D}_{\text{val}} = \{ (x_t, y_t) \mid T_1 < t \le T_2 \}, \quad \mathcal{D}_{\text{test}} = \{ (x_t, y_t) \mid t > T_2 \}$$
